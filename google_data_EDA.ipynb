```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("google_data.csv")
```

```python
obj_cols=[feature for feature in df.columns if df[feature].dtype=='O']
num_cols=[feature for feature in df.columns if df[feature].dtype!='O']
```

```python
for col in obj_cols:
    print(df[col].value_counts(normalize=True)*100)
    print('##############################################')
```

```python
plt.figure(figsize=(15,15))
plt.suptitle('Univariate Analysis of Categorical features', fontsize=20,fontweight='bold',alpha=0.8,y=1.)
category=['Type','Content Rating']
for i in range(0,len(category)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[category[i]],palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```

```python
plt.figure(figsize=(15,15))
plt.suptitle('Univariate Analysisi of Numerical features', fontsize=20,fontweight='bold',alpha=0.8,y=1.)
for i in range(0,len(num_cols)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(x=df[num_cols[i]],shade=True,color='r')
    plt.xlabel(num_cols[i])
    plt.tight_layout()
```

## Observation
Rating and Year is left skewed.
Reviews,size,installs and price and right skewed.
